```python
def shiftGrid(grid, k):
    for e in range(k):
        t = []
        for f in grid:
            t.append(f[-1])
            f[:] = [f[-1]] + f[:-1]
            #注意for f in grid，这里的f只是一个指针
            #如果只写f = [f[-1]] + f[:-1]只是修改了指针的内容，并没有修改原内容
        t = [t[-1]] + t[:-1]
        for i in range(len(grid)):
            grid[i] = [t[i]] + grid[i][1:]
    return grid
shiftGrid(grid = [[1,2,3],[4,5,6],[7,8,9]], k = 1)
```
